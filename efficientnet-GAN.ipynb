# Seed Quality Classification + XAI
**EfficientNet-B2 | `broken`, `discolored`, `pure`, `silkcut`

## Cell 1 — Imports and configuration

```python
import os, time, copy, random
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image
import torch
from torch import nn
from torch.utils.data import DataLoader, ConcatDataset, WeightedRandomSampler
from torchvision import datasets, transforms, models
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, roc_curve, auc
from sklearn.preprocessing import label_binarize

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
if device.type == "cuda":
    torch.cuda.manual_seed_all(SEED)
print("Device:", device)
if device.type == "cuda":
    print("GPU:", torch.cuda.get_device_name(0))

TRAIN_DIR = r"/kaggle/input/datasets/bmuthukumar07/seed-data/data/train"
VAL_DIR = r"/kaggle/input/datasets/bmuthukumar07/seed-data/data/val"
GAN_DIR = r"/kaggle/input/datasets/bmuthukumar07/gan-data-1000-per-class"
GAN_IMAGES_PER_CLASS = 1000

IMG_SIZE = 224
BATCH_SIZE = 32
NUM_WORKERS = 2
EPOCHS = 15
PATIENCE = 5
LR = 1e-4


print("Train directory exists:", os.path.isdir(TRAIN_DIR))
print("Validation directory exists:", os.path.isdir(VAL_DIR))
print("GAN directory exists:", os.path.isdir(GAN_DIR))
```

## Cell 2 — Image transforms

```python
normalize = transforms.Normalize(
    [0.485, 0.456, 0.406],
    [0.229, 0.224, 0.225]
)

train_tfms = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    transforms.ColorJitter(
        brightness=0.10,
        contrast=0.10,
        saturation=0.08
    ),
    transforms.ToTensor(),
    normalize
])

eval_tfms = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    normalize
])

print("Transforms ready.")
```

## Cell 3 — Load train and validation datasets

```python
# Load real training, validation, and existing GAN-generated datasets
train_real = datasets.ImageFolder(TRAIN_DIR, transform=train_tfms)
val_ds = datasets.ImageFolder(VAL_DIR, transform=eval_tfms)
train_gan = datasets.ImageFolder(GAN_DIR, transform=train_tfms)

class_names = train_real.classes
num_classes = len(class_names)

assert num_classes == 4, f"Expected 4 classes, found {class_names}"
assert val_ds.classes == class_names, (
    f"Train classes {class_names} differ from validation classes {val_ds.classes}"
)
assert train_gan.classes == class_names, (
    f"GAN classes {train_gan.classes} differ from real classes {class_names}"
)

# Select exactly 1,000 GAN images from each class.
# The GAN folder must contain one subfolder per class, matching TRAIN_DIR.
random.seed(SEED)
gan_indices = []

for class_idx, class_name in enumerate(class_names):
    class_indices = [
        idx for idx, (_, label) in enumerate(train_gan.samples)
        if label == class_idx
    ]

    if len(class_indices) < GAN_IMAGES_PER_CLASS:
        raise ValueError(
            f"Class '{class_name}' has {len(class_indices)} GAN images; "
            f"{GAN_IMAGES_PER_CLASS} are required."
        )

    random.shuffle(class_indices)
    selected_indices = class_indices[:GAN_IMAGES_PER_CLASS]
    gan_indices.extend(selected_indices)

    print(f"{class_name}: selected {len(selected_indices)} GAN images")

gan_selected = torch.utils.data.Subset(train_gan, gan_indices)

# Combine real and GAN images for TRAINING ONLY.
train_ds = ConcatDataset([train_real, gan_selected])

print("\nClasses:", class_names)
print("Real training images:", len(train_real))
print("GAN images selected:", len(gan_selected))
print("Combined training images:", len(train_ds))
print("Validation images (real only):", len(val_ds))

print("\nPer-class dataset counts:")
for class_idx, class_name in enumerate(class_names):
    real_count = sum(label == class_idx for label in train_real.targets)
    gan_count = sum(
        train_gan.targets[idx] == class_idx for idx in gan_indices
    )
    print(
        f"{class_name}: Real={real_count}, GAN={gan_count}, "
        f"Combined={real_count + gan_count}"
    )
```

## Cell 4 — Create DataLoaders (single GPU)

```python

train_loader = DataLoader(
    train_ds, batch_size=BATCH_SIZE, shuffle=True,
    num_workers=NUM_WORKERS, pin_memory=(device.type == "cuda"),
    persistent_workers=(NUM_WORKERS > 0)
)

val_loader = DataLoader(
    val_ds, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=NUM_WORKERS, pin_memory=(device.type == "cuda"),
    persistent_workers=(NUM_WORKERS > 0)
)

print("Train batches:", len(train_loader))
print("Validation batches:", len(val_loader))
```

## Cell 5 — Build EfficientNet-B2

```python
model = models.efficientnet_b2(
    weights=models.EfficientNet_B2_Weights.DEFAULT
)
model.classifier[1] = nn.Linear(
    model.classifier[1].in_features,
    len(class_names)
)

model = model.to(device)

criterion = nn.CrossEntropyLoss(label_smoothing=0.05)

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=1e-4,
    weight_decay=1e-4
)

scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
    optimizer,
    T_max=EPOCHS,
    eta_min=1e-6
)

# 7. Mixed precision
scaler = torch.amp.GradScaler(
    "cuda",
    enabled=(device.type == "cuda")
)

# 8. Verify trainable parameters
trainable = sum(
    p.numel() for p in model.parameters() if p.requires_grad
)
total = sum(p.numel() for p in model.parameters())

print(f"Trainable parameters: {trainable:,} / {total:,}")
print(f"Trainable percentage: {100 * trainable / total:.2f}%")

print("Model ready on:", device)
print("Loss: unweighted CrossEntropyLoss")
print("Checkpoint selection: highest validation accuracy")
```

## Cell 6 — Define one training/validation epoch

```python
def run_epoch(loader, training):
    model.train(training)
    total_loss = 0.0
    all_labels, all_preds = [], []

    with (torch.enable_grad() if training else torch.no_grad()):
        for images, labels in loader:
            images = images.to(device, non_blocking=True)
            labels = labels.to(device, non_blocking=True)

            if training:
                optimizer.zero_grad(set_to_none=True)

            with torch.amp.autocast(
                device_type=device.type,
                enabled=(device.type == "cuda")
            ):
                outputs = model(images)
                loss = criterion(outputs, labels)

            if training:
                scaler.scale(loss).backward()
                scaler.step(optimizer)
                scaler.update()

            total_loss += loss.item() * images.size(0)
            all_labels.extend(labels.detach().cpu().numpy())
            all_preds.extend(outputs.argmax(1).detach().cpu().numpy())

    return total_loss / len(loader.dataset), accuracy_score(all_labels, all_preds)
```

## Cell 7 — Train with early stopping

```python
history = {
    "train_loss": [], "val_loss": [],
    "train_acc": [], "val_acc": []
}
best_val_acc = -1.0
best_state = copy.deepcopy(model.state_dict())
wait = 0
PATIENCE = 5
training_start = time.time()

for epoch in range(1, EPOCHS + 1):
    epoch_start = time.time()
    train_loss, train_acc = run_epoch(train_loader, training=True)
    val_loss, val_acc = run_epoch(val_loader, training=False)
    scheduler.step()
    epoch_time = (time.time() - epoch_start) / 60

    history["train_loss"].append(train_loss)
    history["val_loss"].append(val_loss)
    history["train_acc"].append(train_acc)
    history["val_acc"].append(val_acc)

    print(
        f"Epoch {epoch}/{EPOCHS} | "
        f"Train loss={train_loss:.4f}, acc={train_acc:.4f} | "
        f"Val loss={val_loss:.4f}, acc={val_acc:.4f} | "
        f"Time: {epoch_time:.2f} min"
    )

    # Save best validation-accuracy checkpoint.
    if val_acc > best_val_acc:
        best_val_acc = val_acc
        best_state = copy.deepcopy(model.state_dict())
        torch.save(best_state, "/kaggle/working/best_seed_efficientnet_b2.pth")
        wait = 0
    else:
        wait += 1
        if wait >= PATIENCE:
            print(f"Early stopping after epoch {epoch}.")
            break

model.load_state_dict(best_state)
print(f"Best validation accuracy: {best_val_acc:.4f}")
print(f"Total training time: {(time.time()-training_start)/60:.2f} min")
print("Best checkpoint saved to /kaggle/working/best_seed_efficientnet_b2.pth")
```

## Cell 8 — Plot training curves

```python
epochs_ran = range(1, len(history["train_loss"]) + 1)
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(epochs_ran, history["train_loss"], label="Train")
axes[0].plot(epochs_ran, history["val_loss"], label="Validation")
axes[0].set_title("Loss")
axes[0].set_xlabel("Epoch")
axes[0].legend()

axes[1].plot(epochs_ran, history["train_acc"], label="Train")
axes[1].plot(epochs_ran, history["val_acc"], label="Validation")
axes[1].set_title("Accuracy")
axes[1].set_xlabel("Epoch")
axes[1].legend()

plt.tight_layout()
plt.show()
```

## Cell 9 — Evaluate and plot confusion matrix + ROC

```python
model.load_state_dict(torch.load("/kaggle/working/best_seed_efficientnet_b2.pth", map_location=device, weights_only=True))
model.eval()
y_true, y_pred, y_prob = [], [], []

with torch.no_grad():
    for images, labels in val_loader:
        images = images.to(device, non_blocking=True)
        with torch.amp.autocast(
            device_type=device.type,
            enabled=(device.type == "cuda")
        ):
            logits = model(images)
        probs = torch.softmax(logits, dim=1)
        y_true.extend(labels.numpy())
        y_pred.extend(probs.argmax(1).cpu().numpy())
        y_prob.extend(probs.cpu().numpy())

y_true = np.array(y_true)
y_pred = np.array(y_pred)
y_prob = np.array(y_prob)

print("Validation accuracy:", round(accuracy_score(y_true, y_pred), 4))
print(classification_report(
    y_true, y_pred, target_names=class_names, digits=4, zero_division=0
))

cm = confusion_matrix(y_true, y_pred, labels=np.arange(num_classes))
plt.figure(figsize=(7, 6))
sns.heatmap(
    cm, annot=True, fmt="d", cmap="Blues",
    xticklabels=class_names, yticklabels=class_names
)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.tight_layout()
plt.show()

y_bin = label_binarize(y_true, classes=np.arange(num_classes))
plt.figure(figsize=(8, 6))
for i, name in enumerate(class_names):
    fpr, tpr, _ = roc_curve(y_bin[:, i], y_prob[:, i])
    plt.plot(fpr, tpr, label=f"{name} (AUC={auc(fpr, tpr):.3f})")
plt.plot([0, 1], [0, 1], "k--", label="Random")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("One-vs-Rest ROC Curves")
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

## Cell 10 — Grad-CAM XAI

```python
# Grad-CAM XAI: Display images from all 4 classes

import random
import torch
import numpy as np
import matplotlib.pyplot as plt

target_layer = model.features[-1]
cam_cache = {"activations": None, "gradients": None}

def save_activation(module, inputs, output):
    cam_cache["activations"] = output

def save_gradient(module, grad_input, grad_output):
    cam_cache["gradients"] = grad_output[0]

hook1 = target_layer.register_forward_hook(save_activation)
hook2 = target_layer.register_full_backward_hook(save_gradient)


def denormalize(tensor):
    mean = torch.tensor(
        [0.485, 0.456, 0.406], device=tensor.device
    ).view(3, 1, 1)

    std = torch.tensor(
        [0.229, 0.224, 0.225], device=tensor.device
    ).view(3, 1, 1)

    return (tensor * std + mean).clamp(0, 1)


def make_gradcam(image_tensor):
    model.eval()
    x = image_tensor.unsqueeze(0).to(device)
    model.zero_grad(set_to_none=True)

    with torch.enable_grad():
        logits = model(x)
        predicted_class = int(logits.argmax(dim=1).item())
        logits[0, predicted_class].backward()

    activations = cam_cache["activations"].detach()
    gradients = cam_cache["gradients"].detach()

    weights = gradients.mean(dim=(2, 3), keepdim=True)
    cam = torch.relu(
        (weights * activations).sum(dim=1, keepdim=True)
    )

    cam = torch.nn.functional.interpolate(
        cam,
        size=(IMG_SIZE, IMG_SIZE),
        mode="bilinear",
        align_corners=False
    )[0, 0]

    cam = cam.cpu().numpy()
    cam = (cam - cam.min()) / (cam.max() - cam.min() + 1e-8)

    probabilities = torch.softmax(
        logits.detach(), dim=1
    )[0].cpu().numpy()

    return cam, predicted_class, probabilities


# Select one validation image from each class
random.seed(42)
selected_indices = []

for class_idx, class_name in enumerate(class_names):
    indices = [
        i for i, (_, label) in enumerate(val_ds)
        if label == class_idx
    ]

    if not indices:
        print(f"No validation images found for {class_name}")
        continue

    selected_indices.append(random.choice(indices))


# Plot original, heatmap, and overlay
fig, axes = plt.subplots(
    len(selected_indices), 3,
    figsize=(13, 4 * len(selected_indices))
)

if len(selected_indices) == 1:
    axes = np.expand_dims(axes, axis=0)

try:
    for row, idx in enumerate(selected_indices):
        image_tensor, true_label = val_ds[idx]

        cam, pred_label, probs = make_gradcam(image_tensor)

        image = denormalize(
            image_tensor.to(device)
        ).permute(1, 2, 0).cpu().numpy()

        axes[row, 0].imshow(image)
        axes[row, 0].set_title(
            f"Original\nTrue: {class_names[true_label]}"
        )

        axes[row, 1].imshow(cam, cmap="jet", vmin=0, vmax=1)
        axes[row, 1].set_title("Grad-CAM Heatmap")

        axes[row, 2].imshow(image)
        axes[row, 2].imshow(
            cam, cmap="jet", alpha=0.4, vmin=0, vmax=1
        )
        axes[row, 2].set_title(
            f"Overlay\nPredicted: {class_names[pred_label]} "
            f"({probs[pred_label]:.2%})"
        )

        for ax in axes[row]:
            ax.axis("off")

    plt.tight_layout()
    plt.show()

finally:
    hook1.remove()
    hook2.remove()
    model.zero_grad(set_to_none=True)
```

## Cell 11 — Visualize misclassified images

```python
wrong_indices = np.where(y_true != y_pred)[0]
print("Misclassified:", len(wrong_indices), "of", len(y_true))

n_show = min(8, len(wrong_indices))
if n_show:
    fig, axes = plt.subplots(2, 4, figsize=(14, 7))
    axes = axes.flatten()
    for ax in axes:
        ax.axis("off")
    for ax, idx in zip(axes, wrong_indices[:n_show]):
        image_tensor, true_label = val_ds[int(idx)]
        image = denormalize(image_tensor.to(device)).permute(1, 2, 0).cpu().numpy()
        ax.imshow(image)
        ax.set_title(f"True: {class_names[true_label]}\\nPred: {class_names[y_pred[idx]]}")
        ax.axis("off")
    plt.tight_layout()
    plt.show()
```

## Cell 13 — Predict a new image (optional)

```python
def predict_image(image_path):
    image = Image.open(image_path).convert("RGB")
    tensor = eval_tfms(image).unsqueeze(0).to(device)
    model.eval()
    with torch.no_grad():
        with torch.amp.autocast(
            device_type=device.type,
            enabled=(device.type == "cuda")
        ):
            logits = model(tensor)
        probs = torch.softmax(logits, dim=1)[0].cpu().numpy()

    order = np.argsort(probs)[::-1]
    for idx in order:
        print(f"{class_names[idx]}: {probs[idx]:.4f}")

    plt.figure(figsize=(4, 4))
    plt.imshow(image.resize((IMG_SIZE, IMG_SIZE)))
    plt.title(f"Prediction: {class_names[order[0]]} ({probs[order[0]]:.2%})")
    plt.axis("off")
    plt.show()
    return class_names[order[0]], probs

# Example:
# predict_image("/kaggle/input/your-dataset/example.jpg")
```
